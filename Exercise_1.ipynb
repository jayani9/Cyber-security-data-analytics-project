## Pandas Series - selecting values, descriptive statistics and filtering

In the next cell, a Pandas series from the list "severity_level" has been created. The index labels have been set to dates with times.

a) Select the last two values from the series using index numbers.

b) Select the values from the series for the time interval 2024-01-02 08:00:00 - 2024-01-02 11:00:00 using index labels.

c) What unique values appear in the series? (Print out all the unique values)

d) Select all rows where the value is "High".

e) How many "High" or "Medium" values are there in the series?

Scoring: 0.5 points for each part, total 2.5 points.

```python
import pandas as pd

# A list of data
severity_level = ['Low', 'Low', 'Low', 'Medium', 'Low', 'Medium', 'High', 'High', 'High', 'Medium', 'Medium', 'Low', 'Medium', 'Low', 'High',
       'Medium', 'High', 'Low', 'High']

# Creating indexes
indexes = pd.date_range(start='1/2/2024', freq='H', periods=19)

# Creating a series
ser = pd.Series(data=severity_level, index=indexes)
```

```python
# Cell /solution 1.1 a)
# Select the last two values from the series using index numbers. Copy and paste the whole output to Moodle's automatic reviewing.

ser.tail(2)
```

```python
# Cell /answer 1.1 b)
# Select values from the series for the time interval 2024-01-02 08:00:00 - 2024-01-02 11:00:00 using index labels.

ser['2024-01-02 08:00:00':'2024-01-02 11:00:00']
```

```python
# Cell /answer 1.1 c)
# What unique values appear in the series? (Print out all the unique values that the series contains). Find out the function specific function that you can use to print out unique values.
ser.unique()
```

```python
# Cell /answer 1.1 d)
# Select all rows where the value is "High".
ser[ser == 'High']
```

```python
# Cell /answer 1.1 e)
# How many "High" or "Medium" values are there in the series?
se = ser[ser.isin(['High', 'Medium'])].count()
print(se)
```

Below are two lists, where the list "anomaly_score" contains the numerical value of network traffic anomalies. The list "severity_level" classifies the values of the "anomaly_score" list into three categories: 'Low', 'Medium', and 'High', which describe the level of traffic anomaly. Create two Series from the lists (named, for example, sl and ans). You can think of having two columns of a table, from which you can filter data using only one column or both columns. Determine the following by filtering and using built-in functions:

a) How many elements in the Anomaly Scores series are over 50?

b) How many elements have an Anomaly Score over 50 and a Severity Level of "High" or "Medium"?

c) What is the average value of the Anomaly Scores?

d) What is the maximum value of the non-"High" values (sl is not "High")?

e) Select the first 3 characters of the values at indices 3 and 4 in the Severity Level series. Note that indexing starts at zero.

Avoid using for-functions, use slicing instead!

Scoring: 0.5 points for each part, total 2.5 points.

```python
anomaly_scores = [28.67, 51.5 , 87.42, 15.79,  0.52,  5.76, 31.55, 54.05, 56.34,
       16.51, 24.91, 86.07, 74.2 , 62.14, 72.65, 91.56, 43.83, 40.46, 44.75]

severity_level = ['Low', 'Low', 'Low', 'Medium', 'Low', 'Medium', 'High', 'High', 'High', 'Medium', 'Medium', 'Low', 'Medium', 'Low', 'High',
       'Medium', 'High', 'Low', 'High']

sl = pd.Series(severity_level) # Severity Level -series
ans = pd.Series(anomaly_scores) # Anomaly Scores -series
```

```python
# Cell / answer 1.2 a)
# How many elements in the Anomaly Scores series are over 50?
an = ans[ans > 50].count()
print(an)
```

```python
# Cell / answer 1.2 b)
# How many elements have an Anomaly Score over 50 and a Severity Level of "High" or "Medium"?
anse = ans[(ans > 50) & (sl.isin(['High', 'Medium']))].count()
print(anse)
```

```python
# Cell / answer 1.2 c)
# What is the average value of the Anomaly Scores? Use a in-build function to determine the average.
avg = ans.mean()
print(avg)
```

```python
# Cell / answer 1.2 d)
# What is the maximum value of the non-"High" values (sl is not "High")? Use the combination of filtering and in-build functions.
ans[sl != "High"].max()
```
