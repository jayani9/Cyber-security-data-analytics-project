```python
# Read the geo-information
df_geo = pd.read_csv('https://student.labranet.jamk.fi/~ajota/Datasets/geoip2-ipv4.csv')
```

```python
# Codes in this cell...
import numpy as np
list1 = df_geo["network"].tolist()
networks = [ipaddress.ip_network(net) for net in list1]
def find_network(ip):
    for net in networks:
        if ip in net:
            return net
    return np.nan
df_m["network"] = df_m["Source IP Address type(ip)"].apply(find_network)
df_m.head(5)
```
