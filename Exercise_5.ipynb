## Timestamps - Part 2


Sometimes, datasets represent time as Unix timestamps, where time is given in seconds since January 1, 1970, 00:00:00 UTC.

The goal of this exercise is to learn how to convert Unix time to Pandas Timestamp for easier readability and handling.

a) Read the 'normal_traffic_unsw_nb15.csv' data into a dataframe from the address
'https://student.labranet.jamk.fi/~ajota/Datasets/normal_traffic_unsw_nb15.csv'.

b) The data columns contain the information 'record start time'. Create a new column 'Timestamp', where the 'record start time' column is converted from Unix time to Pandas Timestamp. At this stage, make the timestamps tz-naive. Do not add timezone information yet. Sort the dataframe by timestamp in ascending order (the earliest timestamp first), use inplace=True for sorting, and reset the index. Print the first entry of the column.

c) Create new columns 'Year', 'Month', 'Day', and 'Hour', where you extract the corresponding year, month, day, and hour information from the 'Timestamp' column. (Note that the 'Timestamp' column data are (or should be!) Pandas Timestamp objects. Check the Pandas documentation to see what attributes the Timestamp object has to easily create the new columns.) Print the 'Timestamp', 'Year', 'Month', 'Day', and 'Hour' information from the first row of the dataframe.

d) Create a new column 'Timestamp Canberra', where the 'Timestamp' column data are localized to the Australian Canberra timezone.

e) Create a new column 'Timestamp Helsinki', where the 'Timestamp Canberra' column data are converted to the Helsinki timezone.

f) Use the Hour column to calculate which hour has the fewest rows/entries in the data.

Scoring: 3 p

```python
# Reading of the data
dfn = pd.read_csv('https://student.labranet.jamk.fi/~ajota/Datasets/normal_traffic_unsw_nb15.csv', index_col=0)
dfn
```

```python
# Cell / answer 1.2 b)
import pandas as pd
# Write the code below...

dfn["Timestamp"] = pd.to_datetime(dfn["record start time"],unit="s",origin="unix")

dfn.sort_values(by="Timestamp",ascending=True,inplace=True)
dfn.reset_index(inplace=True)
# Leave this last and copy the output to Moodle.
dfn['Timestamp'][0]
```

```python
# Cell / answer 1.2 c)

# Write the code below...
dfn["Year"] = dfn["Timestamp"].dt.year
dfn["Month"] = dfn["Timestamp"].dt.month
dfn["Day"] = dfn["Timestamp"].dt.day
dfn["Hour"] = dfn["Timestamp"].dt.hour
dfn[['Timestamp', 'Year', 'Month', 'Day', 'Hour']].head(1)
# Leave this last and copy the output to Moodle.
```

```python
# Cell / answer 1.2 d)


# Write the code below...
dfn['Timestamp Canberra'] = dfn["Timestamp"].dt.tz_localize(tz="Australia/Canberra")

# Leave this last and copy the output to Moodle.
dfn['Timestamp Canberra'][0]
```

```python
# Cell / answer 1.2 e)

# Write the code below...
dfn['Timestamp Helsinki'] = dfn["Timestamp Canberra"].dt.tz_convert(tz="Europe/Helsinki")


# Leave this last and copy the output to Moodle.
dfn['Timestamp Helsinki'][0]
```

```python
# Cell / answer 1.2 f)
# Which hour has the fewest records? Write the corresponding hour to Moodle.
dfn["Hour"].value_counts().sort_values(ascending=True)
# 11
```

## Timestamps - Part 3

In the following tasks, we will use the data from the previous section. Find the answers to the following questions by resampling the data.

a) From which IP address were there the most connections on an hourly basis?

b) When does the data show the most connections (rows) when viewed in 30-minute intervals? Use the 'Timestamp' column and the resample function to find this information.

Scoring: 1 p

```python
# Cell / answer 1.3 a)
# Write in Moodle the 'Source IP address' from which there have been the most connections on an hourly basis.
dfn["Hour"] = dfn["Timestamp"].dt.floor("h")
res = dfn.groupby(["Hour","Source IP address"]).size().reset_index(name="connections")
res = res.loc[res.groupby("Hour")["connections"].idxmax()]
res.sort_values("connections")

```

```python
# Cell / answer 1.3 b)
# Write in Moodle the timestamp when there were the most connections. (In fact, it is a time interval, from which only the first value of the interval is shown in the summary. Sampled at 30-minute intervals, the first times are 2015-01-22 11:30:00, 2015-01-22 12:00:00, 2015-01-22 12:30:00, ...)
dfn = dfn.set_index("Timestamp")
resampled = dfn.resample("30min").size()
resampled.idxmax()
```
