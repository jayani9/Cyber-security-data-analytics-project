## REST API

Fetch the IC1 train data from the Digitraffic API for the date 17.10.2023. The URL to use for the query is 'https://rata.digitraffic.fi/api/v1/trains/2023-10-17/1'. The query returns a JSON object containing nested structures. Create a DataFrame from the fetched data, where the timetable rows (timeTableRows) are parsed into rows of the DataFrame. Set the metadata as meta = ['trainNumber', 'departureDate', 'trainType']. Note that the timeTableRows data also includes the train number 'trainNumber', so you need to set the 'record_prefix' parameter in the JSON normalization (e.g., use the prefix 'tTR').

Finally, print the first 5 rows of the DataFrame. The parsed DataFrame should have the columns (tTR_stationShortCode, tTR_stationUICCode, tTR_countryCode, ..., trainNumber, departureDate, trainType).

```python
import requests
import pandas as pd
```

```python
# Answer
url = "https://rata.digitraffic.fi/api/v1/trains/2023-10-17/1"
data = requests.get(url).json()   # get JSON data

df = pd.json_normalize(data,record_path=["timeTableRows"],meta=["trainNumber","departureDate","trainType"],record_prefix="tTR")      # convert to DataFrame

df = pd.DataFrame(df)
df.head(5)
```
