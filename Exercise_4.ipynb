```python
import pandas as pd

df_2022 = pd.read_csv('https://student.labranet.jamk.fi/~ajota/Datasets/cybersecurity_attacks_2022.csv', index_col=0)
df_2023 = pd.read_csv('https://student.labranet.jamk.fi/~ajota/Datasets/cybersecurity_attacks_2023_v2.csv', index_col=0)
```

```python
# Cell / answer 1.1 a) - merging of data
# Print the length/number of rows of the merged DataFrame and copy it into the Moodle answer field.



df_2022_2023 = pd.concat([df_2022, df_2023], ignore_index=True)

len(df_2022_2023)
```

## Merging Data - Part 2

Add information about the continent and country name of IP addresses (continent_name and country_name data from the DataFrame df_geo) to the DataFrame df_2022_2023 modified in the previous exercise.

Add information for both the source IP and the destination IP. The goal is to know the traffic flow from source to destination.
The DataFrames df_2023 and df_2022 already have columns Source Network and Destination Network, which you can use for merging.
If the merge results in duplicate columns, drop these.
Name the new columns as follows: Source IP Continent, Source IP Country, Destination IP Continent, and Destination IP Country.
Print the "Timestamp", "Source IP Address", "Destination IP Address", "Source Network", "Destination Network", "Source IP Continent", "Source IP Country", "Destination IP Continent", "Destination IP Country" information.

Below is an example of the desired outcome.

Scoring: 1 p

![malli](https://student.labranet.jamk.fi/~ajota/Data-analytics/images/lopputulos_cybersecurity_attacks_df_2022_2023.png)

```python
df_geo = pd.read_csv('https://student.labranet.jamk.fi/~ajota/Datasets/geoip2-ipv4.csv')
```

```python
# Cell / Answer 1.2
# Copy the country information from the last row of the 'Destination IP Country' column into the Moodle answer field.
geo = df_geo[["network", "continent_name", "country_name"]]
df_2022_2023 = df_2022_2023.merge(
    geo,
    left_on="Source Network",
    right_on="network",
    how="left"
)
df_2022_2023.rename(columns={
    "continent_name": "Source IP Continent",
    "country_name": "Source IP Country"
}, inplace=True)

df_2022_2023.drop(columns=["network"], inplace=True)
df_2022_2023 = df_2022_2023.merge(
    geo,
    left_on="Destination Network",
    right_on="network",
    how="left"
)
df_2022_2023.rename(columns={
    "continent_name": "Destination IP Continent",
    "country_name": "Destination IP Country"
}, inplace=True)

df_2022_2023.drop(columns=["network"], inplace=True)
df_2022_2023.drop_duplicates(inplace=True)
df_2022_2023 = pd.DataFrame(df_2022_2023[["Timestamp","Source IP Address","Destination IP Address","Source Network","Destination Network","Source IP Continent","Source IP Country","Destination IP Continent","Destination IP Country"]])
df_2022_2023.iloc[-1]["Destination IP Country"]
```

```python
# This will be checked in Moodle in two parts:
# Cell / Answer 1.3 b) - continent
# Cell / Answer 1.3 b) - attack type

# b) Group the data first by the destination IP address continent (Destination IP Continent) and then by the attack type (Attack Type) in descending order. Which continent and attack type are first in the grouping?

df_2022_2023.groupby(
    ["Destination IP Continent", "Attack Type"]
).size().sort_values(ascending=False).head(1)
```

```python
# Cell / answer 1.3 c)
# This will be checked in Moodle in two parts: answer 1.3 c) - continent
# This will be checked in Moodle in two parts: answer 1.3 c) - country name
# Group by Source IP Continent and Source IP Country
grouped_c = (
    df_2022_2023
    .groupby(["Source IP Continent", "Source IP Country"])
    .count()
)

# Sort by continent ascending and Source IP Address descending
sorted_c = grouped_c.sort_values(
    by=["Source IP Continent", "Source IP Address"],
    ascending=[True, False]
)
sorted_c.tail(1)

```
